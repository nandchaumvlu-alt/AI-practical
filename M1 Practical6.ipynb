```python
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination
# Step 1: Create Bayesian Network structure
model = DiscreteBayesianNetwork([
    ("Burglary", "Alarm"),
    ("Earthquake", "Alarm"),
    ("Alarm", "JohnCalls"),
    ("Alarm", "MaryCalls")
])
# Step 2: Define Conditional Probability Tables (CPDs)
# P(Burglary)
cpd_burglary = TabularCPD(
    variable="Burglary",
    variable_card=2,
    values=[[0.99], [0.01]],
    state_names={"Burglary": ["False", "True"]}
)
# P(Earthquake)
cpd_earthquake = TabularCPD(
    variable="Earthquake",
    variable_card=2,
    values=[[0.998], [0.002]],
    state_names={"Earthquake": ["False", "True"]}
)
# P(Alarm | Burglary, Earthquake)
cpd_alarm = TabularCPD(
    variable="Alarm",
    variable_card=2,
    values=[
        [0.999, 0.71, 0.06, 0.05],
        [0.001, 0.29, 0.94, 0.95]
    ],
    evidence=["Burglary", "Earthquake"],
    evidence_card=[2, 2],
    state_names={
        "Alarm": ["False", "True"],
        "Burglary": ["False", "True"],
        "Earthquake": ["False", "True"]
    }
)
# P(JohnCalls | Alarm)
cpd_john = TabularCPD(
    variable="JohnCalls",
    variable_card=2,
    values=[
        [0.95, 0.10],
        [0.05, 0.90]
    ],
    evidence=["Alarm"],
    evidence_card=[2],
    state_names={
        "JohnCalls": ["False", "True"],
        "Alarm": ["False", "True"]
    }
)
# P(MaryCalls | Alarm)
cpd_mary = TabularCPD(
    variable="MaryCalls",
    variable_card=2,
    values=[
        [0.99, 0.30],
        [0.01, 0.70]
    ],
    evidence=["Alarm"],
    evidence_card=[2],
    state_names={
        "MaryCalls": ["False", "True"],
        "Alarm": ["False", "True"]
    }
)
# Step 3: Add CPDs to model
model.add_cpds(
    cpd_burglary,
    cpd_earthquake,
    cpd_alarm,
    cpd_john,
    cpd_mary
)
# Step 4: Check model correctness
print("Is Bayesian Network valid?", model.check_model())
# Step 5: Perform inference
inference = VariableElimination(model)
# Query: Probability of Burglary when John and Mary both call
result = inference.query(
    variables=["Burglary"],
    evidence={
        "JohnCalls": "True",
        "MaryCalls": "True"
    }
)
print("\nProbability of Burglary given JohnCalls=True and MaryCalls=True:")
print(result)
print("Nandini T006")
```

```python
# Posterior Probability using Bayesian Network
# Real-Life Example: Car Accident Detection System
# Query: P(Accident | Airbag = True)
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination
# Step 1: Create Bayesian Network structure
model = DiscreteBayesianNetwork([
    ("HighSpeed", "Accident"),
    ("Rain", "Accident"),
    ("Accident", "Airbag")
])
# Step 2: Define probability of HighSpeed
cpd_highspeed = TabularCPD(
    variable="HighSpeed",
    variable_card=2,
    values=[[0.70], [0.30]],
    state_names={"HighSpeed": ["False", "True"]}
)
# Step 3: Define probability of Rain
cpd_rain = TabularCPD(
    variable="Rain",
    variable_card=2,
    values=[[0.80], [0.20]],
    state_names={"Rain": ["False", "True"]}
)
# Step 4: Define probability of Accident
# P(Accident | HighSpeed, Rain)
cpd_accident = TabularCPD(
    variable="Accident",
    variable_card=2,
    values=[
        [0.99, 0.90, 0.80, 0.40],  # Accident = False
        [0.01, 0.10, 0.20, 0.60]   # Accident = True
    ],
    evidence=["HighSpeed", "Rain"],
    evidence_card=[2, 2],
    state_names={
        "Accident": ["False", "True"],
        "HighSpeed": ["False", "True"],
        "Rain": ["False", "True"]
    }
)
# Step 5: Define probability of Airbag
# P(Airbag | Accident)
cpd_airbag = TabularCPD(
    variable="Airbag",
    variable_card=2,
    values=[
        [0.99, 0.05],  # Airbag = False
        [0.01, 0.95]   # Airbag = True
    ],
    evidence=["Accident"],
    evidence_card=[2],
    state_names={
        "Airbag": ["False", "True"],
        "Accident": ["False", "True"]
    }
)
# Step 6: Add CPDs to model
model.add_cpds(
    cpd_highspeed,
    cpd_rain,
    cpd_accident,
    cpd_airbag
)
# Step 7: Check whether model is correct
print("Is Bayesian Network valid?", model.check_model())
# Step 8: Perform inference
inference = VariableElimination(model)
result = inference.query(
    variables=["Accident"],
    evidence={"Airbag": "True"}
)
# Step 9: Display Posterior Probability
print("\nPosterior Probability:")
print("P(Accident | Airbag = True)")
print(result)
print("Nandini T006")
```
