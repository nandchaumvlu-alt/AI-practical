```python
# Probability of having disease
P_D = 0.01
# Probability of positive test if disease exists
P_Pos_D = 0.99
# Probability of positive test if disease does not exist
P_Pos_NotD = 0.05
# Probability of not having disease
P_NotD = 1 - P_D
# Bayes Rule
P_D_Pos = (P_D * P_Pos_D) / ((P_Pos_D * P_D) + (P_Pos_NotD * P_NotD))
print("Probability that person actually has disease:")
print(round(P_D_Pos * 100, 2), "%")
print("Nandini T006")
```

```python
# Probability of Blue taxi
P_Blue = 0.15
# Probability of Green taxi
P_Green = 0.85
# Witness says Blue if taxi is Blue
P_SaysBlue_Blue = 0.80
# Witness says Blue if taxi is Green
P_SaysBlue_Green = 0.20
# Bayes Rule
P_Blue_SaysBlue = (P_SaysBlue_Blue * P_Blue) / (
    (P_SaysBlue_Blue * P_Blue) +
    (P_SaysBlue_Green * P_Green)
)
print("Probability that the taxi was actually Blue:")
print(round(P_Blue_SaysBlue * 100,2),"%")
print("Nandini T006")
```

```python
# Probability of knowing Python
P_Python = 0.20
# Probability of knowing Java
P_Java = 0.80
# Probability of placement if Python
P_Placed_Python = 0.90
# Probability of placement if Java
P_Placed_Java = 0.40
# Bayes Rule
P_Python_Placed = (P_Placed_Python * P_Python) / (
    (P_Placed_Python * P_Python) +
    (P_Placed_Java * P_Java)
)
print("Probability that the placed student knows Python:")
print(round(P_Python_Placed * 100,2),"%")
print("Nandini T006")
```
